```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('//content/drive/MyDrive/brain_train.csv')
df1=pd.read_csv('/content/drive/MyDrive/brain_test.csv')
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
```

```python
df=df.drop(columns=['Unnamed: 0.1','Unnamed: 0'])
```

```python
df1=df1.drop(columns=['Unnamed: 0.1','Unnamed: 0'])
```

```python
y_train=df['target']
X_train=df.drop(columns='target')
```

```python
y_test=df1['target']
X_test=df1.drop(columns='target')
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
print(le.classes_)
```

```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Input,Dense,Dropout
```

```python
model=Sequential([Input(shape=(X_train.shape[1],)),
    Dense(512, activation='relu'),
    Dropout(0.2369727754123584),
    Dense(4, activation='softmax')
])
```

```python
from tensorflow.keras.optimizers import Adam

model.compile(
    optimizer=Adam(learning_rate=1.3940809513637057e-05),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
Hyperparameters: {'n_neurons': 512, 'dropout_rate': 0.2369727754123584, 'learning_rate': 1.3940809513637057e-05, 'batch_size': 16, 'n_layers': 1}
```

```python
history = model.fit(
    X_train,
    y_train,
    epochs=15,
    batch_size=16,
    validation_split=0.2
)
```

```python
loss, accuracy = model.evaluate(X_test, y_test)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
y_pred = model.predict(X_test)
y_pred = y_pred.argmax(axis=1)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("Classification Report:")
print(classification_report(
    y_test,
    y_pred,
    target_names=le.classes_
))
```
